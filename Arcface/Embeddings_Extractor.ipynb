```python
from deepface import DeepFace
from mtcnn import MTCNN
from datasets import load_dataset
import numpy as np
from PIL import Image
from tqdm import tqdm
import tensorflow as tf
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report
```

load the dataset

```python
from datasets import load_dataset
ds = load_dataset("tonyassi/celebrity-1000")
```

```python
# Perform split using HuggingFace
split = ds["train"].train_test_split(test_size=0.2, seed=42)

train_ds = split["train"]
test_ds = split["test"]

print("Train size:", len(train_ds))
print("Test size:", len(test_ds))
```

Load Arcface and MTCNN

```python
from deepface import DeepFace

arcface_model = DeepFace.build_model('ArcFace')
```

```python
detector = MTCNN()
```

```python
def extract_face(image, required_size=(112,112)):
    img = np.array(image)
    faces = detector.detect_faces(img)
    if len(faces) == 0:
        return None
    x, y, w, h = faces[0]['box']
    face = img[y:y+h, x:x+w]
    face = Image.fromarray(face).resize(required_size)
    face = np.array(face).astype('float32') / 255.0
    return face

```

get embedding

```python
def get_embedding(face_img):
    if face_img is None:
        return None
    face_img = np.expand_dims(face_img, axis=0)
    embedding = arcface_model.predict(face_img)[0]
    return embedding
```

embedding extractor


```python
if hasattr(arcface_model, 'model'):
    arcface_model = arcface_model.model

X_train, y_train = [], []
X_test, y_test = [], []

for sample in tqdm(train_ds):
    face = extract_face(sample["image"])
    emb = get_embedding(face)
    if emb is not None:
        X_train.append(emb)
        y_train.append(sample["label"])

for sample in tqdm(test_ds):
    face = extract_face(sample["image"])
    emb = get_embedding(face)
    if emb is not None:
        X_test.append(emb)
        y_test.append(sample["label"])

X_train, X_test = np.array(X_train), np.array(X_test)
y_train, y_test = np.array(y_train), np.array(y_test)
```

```python
import pandas as pd

pd.DataFrame(X_train).to_csv('embeddings.csv', index=False, header=False)
pd.DataFrame(y_train).to_csv('labels.csv', index=False, header=False)
print(" Embeddings and labels saved!")
```
